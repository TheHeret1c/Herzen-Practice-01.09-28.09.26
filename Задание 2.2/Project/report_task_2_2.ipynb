## 1. Установка Julia

Julia — высокоуровневый язык программирования для научных и численных вычислений. Он сочетает простоту синтаксиса, близкую к Python, со скоростью, близкой к C, за счёт JIT-компиляции через LLVM.

Дистрибутив Julia был загружен и установлен на компьютер под управлением Windows из Microsoft Store. Установлена версия **Julia 1.13**.

Проверить установку можно, запустив в терминале команду `julia --version` или открыв интерактивную консоль Julia (REPL).



## 2. Подключение Julia к Visual Studio Code

В качестве среды разработки используется **Visual Studio Code** (https://code.visualstudio.com/). Редактор уже был установлен на компьютере ранее, поэтому для работы с Julia потребовалось только добавить в него поддержку языка.

Для этого установлено официальное расширение **Julia** (проект julia-vscode, https://www.julia-vscode.org/). Порядок установки:
1. Открыть в VS Code панель расширений (`Ctrl+Shift+X`).
2. Найти расширение **Julia** (издатель *julialang*) и нажать **Install**.
3. Расширение автоматически находит установленный дистрибутив Julia. Если этого не произошло, путь к `julia.exe` указывается в настройках в параметре `julia.executablePath`.

Расширение добавляет подсветку синтаксиса, автодополнение, встроенный REPL, отладчик и, главное для этой работы, ядро Julia для Jupyter Notebook. Отдельно устанавливать IJulia и Jupyter при этом не требуется.

Справочная статья: https://code.visualstudio.com/docs/languages/julia



## 3. Создание Jupyter Notebook и выбор ядра Julia

1. В VS Code создана папка проекта, а в ней файл с расширением `.ipynb` (этот отчёт).
2. После открытия файла в правом верхнем углу нажата кнопка **Select Kernel**.
3. В списке ядер выбран установленный дистрибутив **Julia 1.13**.

После этого ячейки с кодом выполняются интерпретатором Julia, а текстовые ячейки оформляются в формате Markdown (включая формулы LaTeX).
